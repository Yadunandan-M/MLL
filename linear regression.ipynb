```python
import pandas as pd
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("Housing.csv")
```

```python
df
```

```python
x = df["area"].to_numpy()
y = df["price"].to_numpy()
X_train, X_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42
)
```

```python
plt.scatter(X_train, y_train)

plt.xlabel("area")
plt.ylabel("price")
plt.title("area vs price")
```

The normal equation for linear regression is given by:
$w = (\Phi^T \Phi)^{-1}\Phi^T y$

$w = [c, m]$

$c$ = intercept, $m$ = slope

```python
w_0 = np.ones((X_train.shape[0], 1))

design_matrix = np.column_stack((w_0, X_train))
```

```python
w = np.linalg.inv(design_matrix.T @ design_matrix) @ design_matrix.T @ y_train
c=w[0]
m = w[1]
print("intercept c is ", c)
print("slope m is ", m)
```

```python
plt.scatter(X_train, y_train)
plt.xlabel("area")
plt.ylabel("price")
plt.title("area vs price")

X = np.linspace(0, 16000, 100)
Y = w[1] * X + w[0]

plt.plot(X, Y, color="red")
```

```python
y_pred = (X_test * m ) + c
rmse = np.sqrt(np.mean((y_pred - y_test)**2))
print("the root mean squared error for test set is ", rmse)
```
